```python
%%writefile DiceRolling.cpp
#include<iostream>
using namespace std;
int main{
  cout<<"Hello World";
  return 0;
}
```

```python
!g++ DiceRolling.cpp -o program
```

You have several errors reported by `g++`:

1.  **`DiceRolling.cpp:3:5: error: cannot declare ‘::main’ to be a global variable`**
    *   **Issue:** This is the primary error. The line `int main{` is incorrect. In C++, the `main` function must have parentheses `()` after its name, even if it takes no arguments. It should be `int main() {`.

2.  **`DiceRolling.cpp:4:22: error: expected ‘}’ before ‘;’ token`**
    *   **Issue:** This is a consequence of the first error. Because `main` wasn't correctly declared as a function, the compiler gets confused about the scope and structure, expecting a closing brace where it shouldn't be.

3.  **`DiceRolling.cpp:4:7: error: invalid user-defined conversion from ‘std::basic_ostream<char>’ to ‘int’`**
    *   **Issue:** Another cascading error. The compiler is trying to interpret the `cout` statement in a way that doesn't make sense because of the prior syntax issues.

4.  **`DiceRolling.cpp:5:3: error: expected unqualified-id before ‘return’`**
    *   **Issue:** Again, due to the incorrect function declaration, the `return 0;` statement is not recognized as being within a proper function body.

5.  **`DiceRolling.cpp:6:1: error: expected declaration before ‘}’ token`**
    *   **Issue:** The final closing brace is also misunderstood as the compiler still thinks it's outside a function context.

The main fix is to correct the `main` function signature. Here's the corrected C++ code:

```python
%%writefile DiceRolling.cpp
#include<iostream>
using namespace std;
int main() {
  cout<<"Hello World";
  return 0;
}
```
